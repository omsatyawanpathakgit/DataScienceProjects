```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer()
tokenizer.fit_on_texts([poems_text])
```

```python
len(tokenizer.word_index)
```

```python
input_sequences = []
for sentence in poems_text.split('\n'):
  tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1,len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[:i+1])
```

```python
#Get the maximum-length:
max_len = max([len(x) for x in input_sequences])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences = pad_sequences(input_sequences, maxlen = max_len, padding='pre')
```

```python
padded_input_sequences
```

```python
#Set the input and output variables and apply padding to them:
X = padded_input_sequences[:,:-1]
y = padded_input_sequences[:,-1]

print(X.shape)
print(y.shape)
```

```python
#Set the selected word to 1 and all other words as 0 instead of relying on confidence score:
from tensorflow.keras.utils import to_categorical
y = to_categorical(y,num_classes=649)
```

```python
y.shape
```

```python
#Build the Neural Network:
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

model = Sequential()
model.add(Embedding(649, 100, input_length=12))
model.add(LSTM(150, return_sequences=True))
model.add(LSTM(150))
model.add(Dense(649, activation='softmax'))

model.compile(loss='categorical_crossentropy', optimizer='adam',metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X,y,epochs=150)
```

```python
#Make predictions:
import time
import numpy as np
text = "The shop was closed.".lower()

for i in range(10):
  # tokenize
  token_text = tokenizer.texts_to_sequences([text])[0]
  # padding
  padded_token_text = pad_sequences([token_text], maxlen=56, padding='pre')
  # predict
  pos = np.argmax(model.predict(padded_token_text))

  for word,index in tokenizer.word_index.items():
    if index == pos:
      text = text + " " + word
      print(text)
      time.sleep(2)
```

```python
#Export the model and the tokenizer:
import pickle

with open('flower_poetry_lstm.pkl', 'wb') as f:
    pickle.dump(model, f)

with open('tokenizer.pkl', 'wb') as f:
    pickle.dump(tokenizer, f)
```
